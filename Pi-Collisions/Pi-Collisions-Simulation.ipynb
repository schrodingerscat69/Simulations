```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import matplotlib
matplotlib.use('TkAgg')

# Initialize parameters
m1 = 1      # Mass of Block A
m2 = 1000000 # Mass of Block B
v1 = -1000.0   # Initial velocity of Block A
v2 = 0.0    # Initial velocity of Block B
x1 = 1.0    # Initial position of Block A
x2 = 3.0   # Initial position of Block B
wall_position = 0.0 # Wall at x = 0
dt = 0.001  # Time step
collisions = 0 # Collision counter

# Simulation parameters
block_size = 0.5  # Size of the blocks
fig, ax = plt.subplots(figsize=(10, 4))
ax.set_xlim(-1, 15)
ax.set_ylim(-0.5, 1.5)  # Adjusted ylim for better text positioning
ax.set_aspect('equal')
ax.set_title("Two-Block Collision Simulation")
ax.set_xlabel("Position")
#ax.set_ylabel("Collision Count")
collision_text = ax.text(7, -0.5, '', fontsize=14, color='black', ha='center')  # Adjusted position

block_a = plt.Rectangle((x1, 0), block_size, block_size, color='blue')
block_b = plt.Rectangle((x2, 0), block_size, block_size, color='red')
ax.add_patch(block_a)
ax.add_patch(block_b)

def update(frame):
    global x1, x2, v1, v2, collisions

    # Update positions
    x1 += v1 * dt
    x2 += v2 * dt

    # Check for collision between blocks
    if x1 + block_size >= x2:
        collisions += 1
        # Elastic collision equations
        v1, v2 = ((m1 - m2) * v1 + 2 * m2 * v2) / (m1 + m2), \
                 ((m2 - m1) * v2 + 2 * m1 * v1) / (m1 + m2)

    # Check for collision with the wall
    if x1 <= wall_position:
        collisions += 1
        v1 = -v1

    # Update block positions in the plot
    block_a.set_xy((x1, 0))
    block_b.set_xy((x2, 0))

    # Update collision text
    collision_text.set_text(f"Collisions: {collisions}")

    return block_a, block_b, collision_text

ani = animation.FuncAnimation(fig, update, frames=10000, interval=10, blit=True)

plt.show()
```
