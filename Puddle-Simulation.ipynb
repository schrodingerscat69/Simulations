```python
import numpy as np
from matplotlib import pyplot as plt
from matplotlib import cm
from matplotlib.animation import FuncAnimation
import matplotlib
matplotlib.use('TkAgg')

# Parameters
Lx, Ly = 10, 10
Nx, Ny = 80, 80
x_vec = np.linspace(0, Lx, Nx)
y_vec = np.linspace(0, Ly, Ny)
dx, dy = x_vec[1] - x_vec[0], y_vec[1] - y_vec[0]
dt = 0.01
Nt = 200
c = 1

# Initialize grid
u = np.zeros((Nt, Nx, Ny))
u[0, Nx // 2, Ny // 2] = 1  # Initial disturbance

# Update using finite difference
for t in range(1, Nt-1):
    u[t+1, 1:-1, 1:-1] = (
        c**2 * dt**2 * (
            (u[t, 2:, 1:-1] - 2 * u[t, 1:-1, 1:-1] + u[t, :-2, 1:-1]) / dx**2 +
            (u[t, 1:-1, 2:] - 2 * u[t, 1:-1, 1:-1] + u[t, 1:-1, :-2]) / dy**2
        ) + 2 * u[t, 1:-1, 1:-1] - u[t-1, 1:-1, 1:-1]
    )

# Animation
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
X, Y = np.meshgrid(x_vec, y_vec)
surf = ax.plot_surface(X, Y, u[0], cmap=cm.Blues)

def update(t):
    ax.clear()
    ax.plot_surface(X, Y, u[t], cmap=cm.Blues)
    ax.set_zlim(-0.5, 1.0)
    ax.set_title(f"Time Step: {t}")

anim = FuncAnimation(fig, update, frames=Nt, interval=50)
plt.show()
```
